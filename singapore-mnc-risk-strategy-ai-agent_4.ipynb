Step 5: Risk Scoring Model
## Step 5: Weighted Risk Scoring Model

To calculate the overall MNC risk score, this project uses a weighted scoring model.

The five main risk factors are:

- Cost pressure
- Talent and labour risk
- Competition and growth risk
- Regulation and compliance risk
- Global strategy risk

The weights are assigned based on their strategic importance for MNCs operating in Singapore.

Formula:

Total Risk Score =
Cost Pressure × 0.25 +
Talent and Labour × 0.20 +
Competition and Growth × 0.18 +
Regulation and Compliance × 0.17 +
Global Strategy × 0.20

Risk bands:

- Low risk: below 45
- Medium risk: 45 to 69.9
- High risk: 70 and above

```python
risk_weights = {
    "cost_pressure": 0.25,
    "talent_labour": 0.20,
    "competition_growth": 0.18,
    "regulation_compliance": 0.17,
    "global_strategy": 0.20
}

def calculate_risk_score(row):
    score = 0
    for feature, weight in risk_weights.items():
        score += row[feature] * weight
    return round(score, 1)

def assign_risk_band(score):
    if score >= 70:
        return "High"
    elif score >= 45:
        return "Medium"
    else:
        return "Low"

df["risk_score"] = df.apply(calculate_risk_score, axis=1)
df["risk_band"] = df["risk_score"].apply(assign_risk_band)

df[["industry", "risk_score", "risk_band"]].sort_values(
    by="risk_score", ascending=False
)
```

Step 8: K-Means Strategic Segmentation
# Machine Learning Segmentation using K-Means

To make the project stronger than a normal dashboard, K-Means clustering is used to group industries based on their risk and growth characteristics.

The model uses the following features:

- Cost pressure
- Talent and labour risk
- Competition and growth risk
- Regulation and compliance risk
- Global strategy risk
- Growth potential
- Automation readiness
- Regional expansion fit

The objective is to identify strategic groups such as:
- High-growth controllable risk industries
- Cost/compliance pressure industries
- Regional scale-out candidates

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

ml_features = [
    "cost_pressure",
    "talent_labour",
    "competition_growth",
    "regulation_compliance",
    "global_strategy",
    "growth_potential",
    "automation_readiness",
    "regional_expansion_fit"
]

X = df[ml_features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X_scaled)

df[["industry", "risk_score", "risk_band", "cluster"]]
```

```python
#Cluster naming
cluster_summary = df.groupby("cluster")[ml_features + ["risk_score"]].mean().round(1)

cluster_summary
```

### Step 8B: Justifying the Number of Clusters (Silhouette Analysis)

Before committing to k=3, it's worth checking whether 3 is actually the best-supported number of clusters, or just a convenient one. The cell below runs K-Means for k=2 through k=6 and reports inertia and silhouette score for each.

```python
from sklearn.metrics import silhouette_score

k_results = []
for k in range(2, 7):
    km_k = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_scaled)
    k_results.append({
        "k": k,
        "inertia": round(km_k.inertia_, 1),
        "silhouette_score": round(silhouette_score(X_scaled, km_k.labels_), 3),
    })

k_results_df = pd.DataFrame(k_results)
k_results_df
```

Silhouette scores across k=2-6 fall in a narrow 0.195-0.211 band -- there is no single sharply-optimal k on this dataset. This is expected: industry risk profiles naturally overlap rather than falling into cleanly separated groups. k=3 is used here for business interpretability (it maps cleanly onto three actionable strategic archetypes -- high-value hub, domestic/consumer-facing, and regional trade/supply-chain), not because it statistically dominates the alternatives. This trade-off is intentional and stated explicitly here rather than left implicit.

Step 11: Monte Carlo Stress Testing
## Monte Carlo Stress Testing

A single risk score does not fully capture future uncertainty. In real business conditions, MNC risk can change due to salary inflation, office rental pressure, foreign labour rules, compliance changes, tariffs, geopolitics, and weaker external demand.

To model this uncertainty, Monte Carlo simulation is used. For each industry, the model generates thousands of possible future risk scenarios by applying random shocks to the main risk factors.

The output estimates:

- Expected stressed risk score
- 10th percentile risk
- 90th percentile risk
- Probability of entering the high-risk zone

```python
def monte_carlo_stress_test(row, simulations=5000, shock_strength=8, seed=42):
    np.random.seed(seed)
    
    risk_features = [
        "cost_pressure",
        "talent_labour",
        "competition_growth",
        "regulation_compliance",
        "global_strategy"
    ]
    
    base_values = row[risk_features].values.astype(float)
    weights = np.array([risk_weights[feature] for feature in risk_features])
    
    # Shared macro shock affects cost, talent, and global strategy together
    shared_macro_shock = np.random.normal(0, shock_strength, size=(simulations, 1))
    
    # Individual shocks for each risk factor
    individual_shocks = np.random.normal(
        0, shock_strength * 0.65, size=(simulations, len(risk_features))
    )
    
    shock_vector = np.array([0.75, 0.55, 0.35, 0.25, 0.85])
    
    simulated_values = base_values + shared_macro_shock * shock_vector + individual_shocks
    simulated_values = np.clip(simulated_values, 0, 100)
    
    simulated_scores = simulated_values @ weights
    
    return pd.Series({
        "expected_stressed_risk": round(simulated_scores.mean(), 1),
        "p10_risk": round(np.percentile(simulated_scores, 10), 1),
        "p90_risk": round(np.percentile(simulated_scores, 90), 1),
        "prob_high_risk": round((simulated_scores >= 70).mean(), 3)
    })

stress_results = df.apply(monte_carlo_stress_test, axis=1)

df = pd.concat([df, stress_results], axis=1)

df[[
    "industry",
    "risk_score",
    "expected_stressed_risk",
    "p10_risk",
    "p90_risk",
    "prob_high_risk",
    "strategic_segment"
]].sort_values("prob_high_risk", ascending=False)
```

Step 13: Explainability - Risk Contribution Analysis
## Explainability using Risk Contribution Analysis

To make the model explainable, each industry's total risk score is broken down into weighted contributions from each risk factor.

This is similar to feature-importance thinking in machine learning. It helps answer:

- Which factor contributes most to the final risk score?
- Is the risk driven by cost, labour, compliance, competition, or global strategy?
- What should the company fix first?

```python
def calculate_risk_contributions(row):
    contributions = {}
    
    for feature, weight in risk_weights.items():
        contributions[feature] = round(row[feature] * weight, 2)
    
    return pd.Series(contributions)

contribution_df = df.apply(calculate_risk_contributions, axis=1)
contribution_df["industry"] = df["industry"]

contribution_df
```

```python
#Top driver
top_driver_rows = []

for index, row in contribution_df.iterrows():
    industry = row["industry"]
    
    driver_values = row.drop("industry").sort_values(ascending=False)
    
    top_driver_rows.append({
        "industry": industry,
        "top_driver_1": driver_values.index[0],
        "top_driver_1_points": driver_values.iloc[0],
        "top_driver_2": driver_values.index[1],
        "top_driver_2_points": driver_values.iloc[1],
        "top_driver_3": driver_values.index[2],
        "top_driver_3_points": driver_values.iloc[2],
    })

top_drivers_df = pd.DataFrame(top_driver_rows)

top_drivers_df
```

visualization cell

```python
contribution_plot = contribution_df.set_index("industry")

plt.figure(figsize=(12, 7))

contribution_plot.plot(
    kind="barh",
    stacked=True,
    figsize=(12, 7),
    colormap="Set2"
)

plt.title("Explainable Risk Score: Weighted Contribution by Factor", fontsize=15, fontweight="bold")
plt.xlabel("Weighted Risk Points")
plt.ylabel("Industry")
plt.legend(title="Risk Factor", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

This explainability layer shows that the model is not a black box. Each final risk score can be traced back to individual weighted risk drivers. For example, finance and fintech is driven strongly by cost pressure and compliance, while logistics is more exposed to global strategy risk.

Step 14: AI Agent Recommendation Engine
## AI Agent Recommendation Engine

This section converts the ML outputs into business strategy recommendations.

The AI agent follows a tool-based workflow:

1. Read the selected industry profile
2. Calculate the risk score and risk band
3. Identify the strategic segment
4. Review Monte Carlo stress-test exposure
5. Identify top risk drivers
6. Generate a recommended action plan

This makes the project an agentic decision-support system rather than a static dashboard.

```python
def generate_strategy_recommendation(row):
    recommendations = []
    
    if row["cost_pressure"] >= 70:
        recommendations.append(
            "Move routine and cost-sensitive work to lower-cost regional shared-service locations while keeping high-value leadership, finance, compliance, and innovation functions in Singapore."
        )
    else:
        recommendations.append(
            "Maintain a lean Singapore operating model using hybrid work, vendor optimisation, and productivity tracking."
        )
    
    if row["talent_labour"] >= 68:
        recommendations.append(
            "Reduce talent pressure through upskilling, university partnerships, internal mobility, and automation of repetitive roles."
        )
    
    if row["regulation_compliance"] >= 70:
        recommendations.append(
            "Strengthen compliance systems covering data privacy, employment rules, tax, sector regulations, internal audit, and board reporting."
        )
    
    if row["global_strategy"] >= 65:
        recommendations.append(
            "Use scenario planning for tariffs, external demand shocks, geopolitics, supply-chain disruption, and headquarters restructuring risk."
        )
    
    if row["growth_potential"] >= 78:
        recommendations.append(
            "Use Singapore as a high-value ASEAN launchpad rather than relying only on the small domestic market."
        )
    
    if row["regional_expansion_fit"] >= 78:
        recommendations.append(
            "Expand regionally through partnerships, distributors, joint ventures, or a Singapore-plus-one operating model."
        )
    
    if row["automation_readiness"] >= 75:
        recommendations.append(
            "Prioritise AI-enabled workflows such as finance reporting, customer support triage, demand forecasting, and compliance monitoring."
        )
    
    return recommendations
```

```python
def agentic_strategy_brief(industry_name):
    row = df[df["industry"] == industry_name].iloc[0]
    
    top_drivers = top_drivers_df[top_drivers_df["industry"] == industry_name].iloc[0]
    
    print("=" * 90)
    print("AGENTIC ML STRATEGY BRIEF")
    print("=" * 90)
    print(f"Industry: {row['industry']}")
    print(f"Risk Score: {row['risk_score']} / 100")
    print(f"Risk Band: {row['risk_band']}")
    print(f"Strategic Segment: {row['strategic_segment']}")
    print(f"Probability of High Risk under Stress: {row['prob_high_risk']:.1%}")
    print()
    
    print("Top Risk Drivers:")
    print(f"1. {top_drivers['top_driver_1']} ({top_drivers['top_driver_1_points']} weighted points)")
    print(f"2. {top_drivers['top_driver_2']} ({top_drivers['top_driver_2_points']} weighted points)")
    print(f"3. {top_drivers['top_driver_3']} ({top_drivers['top_driver_3_points']} weighted points)")
    print()
    
    print("Recommended Action Plan:")
    for i, rec in enumerate(generate_strategy_recommendation(row), start=1):
        print(f"{i}. {rec}")
    
    print()
    print("Final Strategic Recommendation:")
    
    if row["risk_band"] == "High" and row["growth_potential"] >= 78:
        print("Keep Singapore as a high-value strategic hub, but avoid scaling routine operations locally. Expand through regional partnerships and automation.")
    elif row["risk_band"] == "Medium" and row["growth_potential"] >= 78:
        print("Scale selectively from Singapore by focusing on high-value functions and regional expansion.")
    elif row["prob_high_risk"] >= 0.50:
        print("Stabilise cost, compliance, and global exposure before major expansion.")
    else:
        print("Maintain lean Singapore operations and use partnerships for regional growth.")
```

Step 15: Generate Recommendations for All Industries
## Final Strategy Recommendation Table

The agent is now applied to all industries to generate a final decision table. This table can be used by business leaders to compare industries and choose the right Singapore strategy.

```python
def final_strategy(row):
    if row["risk_band"] == "High" and row["growth_potential"] >= 78:
        return "High-value hub only; scale routine work regionally"
    elif row["risk_band"] == "Medium" and row["growth_potential"] >= 78:
        return "Selective Singapore scaling with ASEAN expansion"
    elif row["prob_high_risk"] >= 0.50:
        return "Stabilise cost and compliance before expansion"
    elif row["regional_expansion_fit"] >= 80:
        return "Use Singapore as control hub and expand regionally"
    else:
        return "Maintain lean operations and optimise productivity"

df["final_strategy"] = df.apply(final_strategy, axis=1)

final_table = df[[
    "industry",
    "risk_score",
    "risk_band",
    "prob_high_risk",
    "strategic_segment",
    "final_strategy"
]].sort_values("risk_score", ascending=False)

final_table
```

The best strategy is to keep Singapore as a high-value strategic hub, move routine work to cost-effective regional locations, invest in talent and technology, strengthen compliance, and expand across ASEAN through partnerships.

The ML-based analysis identified finance and fintech as the highest-risk sector due to high cost pressure, talent competition, and strict compliance requirements. Advanced manufacturing also showed high risk due to operating cost and global supply-chain exposure. Technology and AI services showed medium risk but strong growth potential, making it suitable for selective scaling from Singapore.

Overall, this project demonstrates how machine learning, stress testing, explainability, and agent-style recommendation logic can support strategic business decision-making.
